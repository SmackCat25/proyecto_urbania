```python
import pandas as pd
```

```python
path = "C:/Users/maria/PycharmProjects/Urbania_Proyecto/Scrapping"
excel_path = f"{path}/df_limpio.xlsx"
output_path = f"{path}/inmuebles_para_botpress.txt"
```

```python
df = pd.read_excel(excel_path)
```

Lo convertimos en un archivo de tipo txt para ser usado en botpress (gratuito)

```python
with open(output_path, "w", encoding="utf-8") as f:
    for i, row in df.iterrows():
        # Encabezado con ID
        f.write(f"ID: {i+1}\n")
        
        # Campos principales
        f.write(f"Tipo de contrato: {row.get('tipo_contrato','')}\n")
        f.write(f"Tipo de vivienda: {row.get('tipo_vivienda','')}\n")
        f.write(f"Precio: {row.get('precio','')} soles\n")
        f.write(f"Departamento: {row.get('departamento','')}\n")
        f.write(f"Provincia: {row.get('provincia','')}\n")
        f.write(f"Distrito: {row.get('distrito','')}\n")
        f.write(f"Dirección: {row.get('direccion','')}\n")
        f.write(f"Área total: {row.get('area_total','')}\n")
        f.write(f"Área cubierta: {row.get('area_cubierta','')}\n")
        f.write(f"Baños: {row.get('banos','')}\n")
        f.write(f"Medios baños: {row.get('medios_banos','')}\n")
        f.write(f"Cochera: {row.get('cochera','')}\n")
        f.write(f"Dormitorios: {row.get('dormitorio','')}\n")
        f.write(f"Antigüedad: {row.get('antiguedad','')}\n")
        
        # Descripción larga (sin saltos de línea)
        descripcion = str(row.get('descripcion','')).replace("\n", " ")
        f.write(f"Descripción: {descripcion}\n")
        
        # Extras
        f.write(f"URL: {row.get('url','')}\n")
        # f.write(f"Fecha extracción: {row.get('fecha_extraccion','')}\n")
        
        # Separador entre registros
        f.write("---\n\n")
```

```python
print(f"✅ Archivo generado con éxito en:\n{output_path}")
```
